# Arena size for experiment

## Why?
We know density of bats in groups ~ 30 bats in a 9mx5mx7m lxbxh space [1]. But how to translate this to arena size for any given number of bats. 

## Process
We will place 30 points using poisson disk sampling in order to estimate what the most appropriate minimum interbat spacing would be for a 9 by 5 space. 
We will then use this minimum interbat spacing to find arena sizes for variable group sizes.

```python
bat_sizes = np.array([5,10,30,50,75,100,200])
ratio_between_sides = 1.5
function_from_bat_to_arena_size = lambda x : np.round(np.sqrt(x*(1/(ratio_between_sides*(2/3)))), 2)
x = function_from_bat_to_arena_size(bat_sizes)
y = ratio_between_sides*x 

print(f"arena length : {x}")
print(f"arena width : {y}")
print(f"density check : {np.round(bat_sizes/(x*y),3)}")
```

```python
df_arena_sizes = pd.DataFrame()
df_arena_sizes["arena length"] = x+2
df_arena_sizes["arena width"] = y+2
df_arena_sizes["group size"] = bat_sizes
df_arena_sizes.to_csv("arena_sizes_with_group_size.csv")
```
