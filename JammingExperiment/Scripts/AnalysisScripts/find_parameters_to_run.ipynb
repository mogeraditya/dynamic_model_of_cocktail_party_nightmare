```python
import glob 
import os

location = "/home/adityamoger/Documents/csv_storage/sensitivity_analysis"
param_files = glob.glob(os.path.join(location, "*.csv"))
param_files = [f for f in param_files if os.path.isfile(f)]
```

```python
param_files
```

```python
def find_param_id(dir):
    output = ""
    for char in dir:
        if char.isdigit():
            output+=char
    return output
        
param_ids = [find_param_id(f) for f in param_files]
```

```python
param_names_to_delete = ["/home/adityamoger/Documents/GitHub/dynamic_model_of_cocktail_party_nightmare/JammingExperiment/Data/InputData/sensitivity_params/paramset_number_"+f+".json" for f in param_ids]
param_names_to_delete
```

```python
def delete_file(x):
    if os.path.exists(x):
        os.remove(x)
    else:
        print("The file does not exist") 
        
for param_file in param_names_to_delete:
    delete_file(param_file)
```
