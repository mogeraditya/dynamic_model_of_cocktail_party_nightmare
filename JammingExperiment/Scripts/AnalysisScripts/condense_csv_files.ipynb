```python
import pandas as pd
import numpy as np

import glob
import os
```

```python
path_of_files = r"/home/adityamoger/Documents/data_storage/csv_storage"
list_of_file_dirs = glob.glob(os.path.join(path_of_files, "*.csv"))
list_of_file_dirs = [f for f in list_of_file_dirs if os.path.isfile(f)]
list_of_file_dirs, len(list_of_file_dirs)
```

```python
flist = []
for filename in list_of_file_dirs:
    df = pd.read_csv(filename, index_col=False)  # <- HERE
    flist.append(df)

df_out = pd.concat(flist, axis=0, ignore_index=False)
df_out.to_csv("sensitivity_analysis_data.csv")
```
